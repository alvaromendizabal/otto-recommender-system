# OTTO · Timing-feature replication design

This is an **executed design notebook using the previous measured Round 01 result**. There are no Round 02 experiment scores yet. Its purpose is to explain why the 24 timing formulas stay frozen for a new-cohort test.

```python
from pathlib import Path
import json
import plotly.graph_objects as go
import plotly.io as pio
from IPython.display import display
ROOT = Path.cwd()
reference = json.loads((ROOT/'evidence/ROUND01_REFERENCE.json').read_text())
def show(fig):
    display({'application/vnd.plotly.v1+json':json.loads(pio.to_json(fig))},raw=True)
print('Source: independently reviewed Round 01 per-session statistics. No new model fits.')
```

## Previous pilot: timing helped; support did not

The four previous models were scored on the same validation sessions. The combined 48-feature primary hypothesis failed. Timing-only is now a **separately preregistered replication hypothesis**, not a promoted feature block.

```python
names=['control_shared','plus_support','plus_timing','plus_both']
fig=go.Figure(go.Bar(x=names,y=[reference['prior_arms'][n]['weighted_recall_at_20'] for n in names],text=[f"{reference['prior_arms'][n]['weighted_recall_at_20']:.6f}" for n in names],textposition='outside'))
fig.update_layout(title='Previous Round 01 only — matched fitting data',template='plotly_white',yaxis_title='Pooled weighted Recall@20')
show(fig)
```

## One order drove much of the small-sample effect

A gain is not meaningful just because it has many decimal places. The validation sample had only 92 order-denominator units. The plot decomposes the actual net change in hits.

```python
old=reference['prior_arms']['control_shared'];new=reference['prior_arms']['plus_timing']
weights=[.1,.3,.6];contributions=[w*(b-a)/d for w,a,b,d in zip(weights,old['hits'],new['hits'],old['denominators'])]
fig=go.Figure(go.Bar(x=['Clicks','Carts','Orders'],y=contributions,text=[f'{b-a:+d} net hits' for a,b in zip(old['hits'],new['hits'])],textposition='outside'))
fig.update_layout(title='Previous timing gain decomposition — not a Round 02 score',template='plotly_white',yaxis_title='Weighted score contribution')
show(fig)
print('Fraction attributable to orders:',contributions[2]/sum(contributions))
```

## What stays fixed and what changes

**Fixed:** 24 timing equations, 134-column control, 400 candidates, Aug-16 historical graphs, model settings and feature names. **Changed:** 4096 fitting-session IDs, excluding the prior 1024, with new matched control models. It is not a new historical window or untouched holdout.

The timing statistics are an affinity-weighted mean of log(1+hours), and log(1+hours) for the strongest supporting observed item. Ages reference latest observed item occurrence. Empty support and zero age remain the same frozen convention.

Domain evidence: [the third-place OTTO author](https://github.com/TheoViel/kaggle_otto_rs) describes candidate-session similarity aggregation with position, timestamp and action information. That supports investigating timing; it does not prove these particular equations work.
