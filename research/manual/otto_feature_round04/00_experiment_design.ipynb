Round03 did not beat its matched control. These first two charts show **actual completed Round03 measurements**. The third is a feature-catalog count, not performance. Full definitions and availability assumptions are in RESEARCH_PLAN.md. This notebook never starts an experiment.

```python
from pathlib import Path
import json
import plotly.graph_objects as go
from IPython.display import display
ROOT=next((p for p in [Path.cwd(),*Path.cwd().parents] if (p/'demand_features.py').is_file()),Path.home()/'otto_feature_round04')
r=json.loads((ROOT/'evidence/ROUND03_RESULT.json').read_text())
def show(fig):
    display({'application/vnd.plotly.v1+json':json.loads(fig.to_json()),'text/plain':fig.layout.title.text},raw=True)
print('Design only — no raw data or models loaded')
```

## Stop the failed primary, not merely the weaker ablation

```python
arms=['control_shared','mass_missing_aware','mass_zero_ablation']
fig=go.Figure(go.Bar(x=['Control','Recent mass, missing aware','Recent mass, zero filled'],y=[r['arms'][a]['weighted_recall_at_20'] for a in arms],text=[f"{r['arms'][a]['weighted_recall_at_20']:.6f}" for a in arms],textposition='outside'))
fig.update_layout(title='Round03: actual pooled fitting scores — not Kaggle scores',yaxis_title='Weighted Recall@20')
show(fig)
```

## A later-fold improvement did not offset earlier regression

```python
fig=go.Figure()
for a in arms[1:]:
    fig.add_trace(go.Bar(name=a,x=['Earlier fold','Later fold'],y=r['comparisons'][a+'_minus_control_shared']['fold_gains']))
fig.add_hline(y=0,line_dash='dash')
fig.update_layout(title='Round03: actual matched chronological gains',yaxis_title='Gain versus control',barmode='group')
show(fig)
```
