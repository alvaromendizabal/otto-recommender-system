# Round 10 · Worked feature example

A clicked → carted → ordered product and a reversed sequence occur in two synthetic neighbors. A third neighbor clicks the product but does not progress. Item 99 has an order without a source action and therefore has no funnel support. **These are synthetic illustrative events, not new project results.**

```python
from pathlib import Path
import sys, json
import numpy as np
import plotly.graph_objects as go
import plotly.io as pio
from IPython.display import display
ROOT=Path.cwd()
if not (ROOT/'funnel_features.py').is_file(): ROOT=Path.home()/'otto_feature_round10'
sys.path.insert(0,str(ROOT))
from neighbors import group_history
from feature_logic import build,names
rows=np.array([(1,10,10,0,0),(1,80,20,0,1),(1,80,30,1,2),(1,80,40,2,3),
 (2,10,10,0,0),(2,80,20,2,1),(2,80,30,1,2),(2,80,40,0,3),
 (3,10,10,0,0),(3,80,20,0,1),(3,99,30,2,2)],np.int64)
history=group_history(rows,100000000,set())
candidates=np.array([80,99,999])
a,b,diag=build(10,np.array([10]),candidates,[1,2,3],history,{10:3})
assert a.shape==b.shape==(3,24)
def show(fig):
    display({'application/vnd.plotly.v1+json':json.loads(pio.to_json(fig))},raw=True)
print('SYNTHETIC_EXAMPLE_ONLY — not an OTTO model score')
print(json.dumps(diag,indent=2))
```

## Click → cart

```python
fig=go.Figure()
fig.add_trace(go.Bar(name='Ordered progression',x=candidates.astype(str),y=a[:,0]))
fig.add_trace(go.Bar(name='Unordered co-action',x=candidates.astype(str),y=b[:,0]))
fig.update_layout(title='Click → cart: one ordered versus two unordered supporting sessions',barmode='group',xaxis_title='Synthetic candidate product',yaxis_title='log(1 + session support)')
show(fig)
```

## Cart → order

```python
fig=go.Figure()
fig.add_trace(go.Bar(name='Ordered progression',x=candidates.astype(str),y=a[:,8]))
fig.add_trace(go.Bar(name='Unordered co-action',x=candidates.astype(str),y=b[:,8]))
fig.update_layout(title='Cart → order: one ordered versus two unordered supporting sessions',barmode='group',xaxis_title='Synthetic candidate product',yaxis_title='log(1 + session support)')
show(fig)
```

## Click → order

```python
fig=go.Figure()
fig.add_trace(go.Bar(name='Ordered progression',x=candidates.astype(str),y=a[:,16]))
fig.add_trace(go.Bar(name='Unordered co-action',x=candidates.astype(str),y=b[:,16]))
fig.update_layout(title='Click → order: one ordered versus two unordered supporting sessions',barmode='group',xaxis_title='Synthetic candidate product',yaxis_title='log(1 + session support)')
show(fig)
```

## Candidate-specific denominators
The click-source denominator is three sessions, but the cart-source denominator is two. These local conditional intensities are not population conversion probabilities.

```python
fig=go.Figure()
fig.add_trace(go.Bar(name='Ordered',x=['Click to cart','Cart to order','Click to order'],y=a[0,[2,10,18]]))
fig.add_trace(go.Bar(name='Unordered',x=['Click to cart','Cart to order','Click to order'],y=b[0,[2,10,18]]))
fig.update_layout(title='Support / (source sessions + 20)',barmode='group')
show(fig)
```

## Full feature contrast

```python
fig=go.Figure(go.Bar(x=list(names(10)),y=np.abs(a-b).mean(axis=0)))
fig.update_layout(title='24 declared feature differences — synthetic only',height=700,xaxis_tickangle=-65)
show(fig)
print('WORKED_EXAMPLE_COMPLETE: 24 primary + 24 ablation columns; no model training')
```
