## 2. Three observable states
Illustrative values below are algebraic examples, not OTTO experiment predictions. No support, support only from old observations, and recent support are distinct states. The primary encoding preserves that distinction.

```python
import numpy as np
from recent_mass_features import summarize_mass
states = {"No graph evidence": ([[0.]], [0.]),
          "Only old evidence": ([[1.]], [3.]),
          "Recent and old evidence": ([[1.], [3.]], [0., 1.])}
fig = go.Figure()
for label, (weights, ages) in states.items():
    seeds = np.arange(1, len(ages)+1)
    fractions = summarize_mass(weights, seeds, np.array([7]), ages)[0]
    fig.add_trace(go.Scatter(x=[5,30,120], y=fractions.tolist(), name=label, mode="lines+markers"))
fig.update_layout(title="Illustration only: affinity mass within each observation-age window",
                  xaxis_title="Window (minutes)", yaxis_title="Fraction; −1 denotes no evidence")
show(fig)
```
