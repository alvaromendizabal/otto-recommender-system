# 01 — Leakage-aware validation protocol

This notebook documents the frozen temporal validation contract used by the project. It is intentionally a reporting notebook: production split logic remains in the package.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / "reports").exists():
    ROOT = ROOT.parent

def load_json(name):
    return json.loads((ROOT / "reports" / "metrics" / name).read_text())
```

```python
v = load_json("validation_summary.json")
pd.DataFrame({
    "quantity": ["validation sessions", "click labels", "cart label items", "order label items"],
    "count": [v["sessions"], v["click_labels"], v["cart_label_items"], v["order_label_items"]],
})
```

### Why this matters

The validation set is time-separated from the retrieval-training universe. This prevents future behavior from entering candidate generation or learned features. The three objectives retain the competition weighting rather than being collapsed into a generic binary target.
